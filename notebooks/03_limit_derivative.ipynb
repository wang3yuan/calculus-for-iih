## 1. 用數值表看極限

$f(x)=\dfrac{x^2-1}{x-1}$ 在 $x=1$ 沒有定義（分子分母都是 0），但 x 靠近 1 時呢？我們從左右兩邊靠近。

```python
f = lambda x: (x**2 - 1) / (x - 1)

print("從左邊靠近        從右邊靠近")
for h in [0.1, 0.01, 0.001, 0.0001]:
    print(f"f({1 - h:<7}) = {f(1 - h):.4f}    f({1 + h:<7}) = {f(1 + h):.4f}")
```

左右都靠近 **2**，所以極限是 2，雖然 $f(1)$ 根本不存在。**極限問的是「靠近時往哪裡去」，不是「那一點的值」。**

數值表只是猜測，要確定還是得用代數：$\dfrac{(x-1)(x+1)}{x-1}=x+1\to 2$。

## 2. 左右不一樣：極限不存在

$g(x)=\dfrac{|x|}{x}$ 在 0 的左邊都是 −1、右邊都是 +1。

```python
g = lambda x: abs(x) / x

for h in [0.1, 0.001, 0.00001]:
    print(f"g({-h:<8}) = {g(-h):+.1f}     g({h:<7}) = {g(h):+.1f}")
```

左極限 −1、右極限 +1，不相等，所以 $\lim_{x\to0}g(x)$ 不存在。

## 4. 兩個重要的極限

課堂提過 $\dfrac{\sin x}{x}\to1$（$x\to0$）和 $\left(1+\dfrac1n\right)^n\to e$（$n\to\infty$）。代數很難化簡，但電腦一看就知道。

```python
for x in [0.5, 0.1, 0.01, 0.001]:
    print(f"sin({x:<5})/{x:<5} = {np.sin(x) / x:.6f}")
print()
for m in [1, 10, 100, 10000, 1000000]:
    print(f"(1 + 1/{m:<7})^{m:<7} = {(1 + 1 / m)**m:.6f}")
print("e =", round(np.e, 6))
```

## 5. 差商就是割線斜率

導數的定義是

$$f'(a)=\lim_{h\to 0}\frac{f(a+h)-f(a)}{h}$$

分子是函數值的變化量，分母是 x 的變化量，整個分數就是**割線的斜率**。用 $f(x)=x^2+x$ 在 $a=1$ 看看。

```python
f = lambda x: x**2 + x
a = 1
for h in [1, 0.5, 0.1, 0.01, 0.001, 0.0001]:
    print(f"h = {h:<7}  割線斜率 = {(f(a + h) - f(a)) / h:.4f}")
```

h 越小，斜率越靠近 **3**。手算也一樣：

$$\frac{\big[(1+h)^2+(1+h)\big]-2}{h}=\frac{3h+h^2}{h}=3+h\ \to\ 3$$

## 6. 看圖：割線慢慢變成切線

h 越小，割線越貼近綠色的切線。**導數就是切線的斜率。**

```python
x = np.linspace(-1, 3.5, 300)

plt.figure(figsize=(7, 5))
plt.plot(x, f(x), color="#0B2E33", lw=3, label="y = x² + x")
for h, c in [(2, "#E8A33D"), (1, "#D9534F"), (0.3, "#028090")]:
    m = (f(a + h) - f(a)) / h
    plt.plot(x, f(a) + m * (x - a), color=c, lw=1.8, label=f"secant, h = {h}, slope = {m:.1f}")
    plt.plot([a + h], [f(a + h)], "o", color=c)
plt.plot(x, f(a) + 3 * (x - a), color="green", lw=2.5, ls="--", label="tangent, slope = 3")
plt.plot([a], [f(a)], "ko")
plt.xlim(-1, 3.5); plt.ylim(-2, 15)
plt.xlabel("x"); plt.ylabel("y"); plt.grid(alpha=0.3); plt.legend()
plt.show()
```

## 9. 電腦算的導數 vs 公式

用很小的 h 當作導數的近似，和公式 $(x^3)'=3x^2$ 比較。

```python
def numeric_derivative(f, a, h=1e-6):
    "用很小的 h 近似導數（老師提供）。"
    return (f(a + h) - f(a)) / h

g = lambda x: x**3
print("  x    電腦算的   公式 3x²")
for a in [-2, -1, 0, 1, 2, 3]:
    print(f"{a:>3}   {numeric_derivative(g, a):>8.3f}   {3 * a**2:>8}")
```

## 10. 情境：體溫的變化率

某位病人治療後的體溫（示意模型）：$T(t)=36.5+1.2t-0.15t^2$（°C，t 為小時）。導數 $T'(t)$ 是體溫的變化率，單位是 °C／小時。

```python
T = lambda t: 36.5 + 1.2 * t - 0.15 * t**2
t = np.linspace(0, 8, 200)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(t, T(t), color="#028090", lw=3)
ax[0].set_xlabel("t (hours)"); ax[0].set_ylabel("T (°C)"); ax[0].set_title("T(t)"); ax[0].grid(alpha=0.3)
ax[1].plot(t, [numeric_derivative(T, s) for s in t], color="#D9534F", lw=3)
ax[1].axhline(0, color="gray", lw=1)
ax[1].set_xlabel("t (hours)"); ax[1].set_ylabel("T' (°C/hour)"); ax[1].set_title("rate of change T'(t)"); ax[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 11. ✏️ 動手 4：體溫什麼時候最高？

右圖的 $T'$ 在某個時間穿過 0：之前是正的（還在上升）、之後是負的（開始下降），那一刻就是最高點。請看圖填入時間。

```python
print("電腦算出 T′(2) ≈", round(numeric_derivative(T, 2), 3), "°C/小時")

answer_4 = None     # ✏️ 填空：體溫最高的時刻（小時）

# ---- 自動檢查（不用改）----
if answer_4 is None:
    print("還沒填，看右圖 T′ 穿過 0 的時間是多少？")
else:
    check(abs(answer_4 - 4) < 0.2, "T′ 從正變負的那個時間就是答案。")
    print(f"那一刻的體溫是 {T(answer_4):.1f} °C")
```

## 12. ✏️ 動手 5：沒有公式，只有數據

連續血糖監測每 5 分鐘一筆，沒有公式，也能估變化率。用**中央差分**：

$$G'(t)\approx\frac{G(t+5)-G(t-5)}{10}$$

下面的資料是**虛擬**的。請估計第 30 分鐘的變化率。

```python
rng = np.random.default_rng(11)
minutes = np.arange(0, 65, 5)
glucose = (110 + 50 * np.sin((minutes - 10) / 20) + rng.normal(0, 1.5, minutes.size)).round(0)

plt.figure(figsize=(7, 3.2))
plt.plot(minutes, glucose, "o-", color="#028090")
plt.xlabel("t (minutes)"); plt.ylabel("glucose (mg/dL)"); plt.grid(alpha=0.3); plt.show()

G = dict(zip(minutes.tolist(), glucose.tolist()))     # 用 G[30] 可以取出第 30 分鐘的血糖
print("G[25] =", G[25], "  G[30] =", G[30], "  G[35] =", G[35])

answer_5 = None     # ✏️ 填空：用 G[35]、G[25] 寫出中央差分

# ---- 自動檢查（不用改）----
if answer_5 is None:
    print("還沒填，把 None 換成 (G[35] - G[25]) / 10 這樣的式子。")
else:
    check(abs(answer_5 - (G[35] - G[25]) / 10) < 1e-9, "分子是後一點減前一點，分母是兩點相差的 10 分鐘。")
    print(f"第 30 分鐘血糖約以每分鐘 {abs(answer_5):.1f} mg/dL 的速度{'上升' if answer_5 > 0 else '下降'}")
```

## 14. h 越小越好嗎？

數學上 h 越小越準，但電腦存小數有誤差。

```python
f = lambda x: x**2 + x
print("   h        割線斜率   誤差（和 3 相差多少）")
for k in range(1, 17):
    h = 10.0**(-k)
    s = (f(1 + h) - f(1)) / h
    print(f"1e-{k:<2}  {s:>12.6f}   {abs(s - 3):.2e}")
```

一開始 h 變小，誤差也變小；但 h 小到大約 $10^{-8}$ 以後，誤差反而變大，最後甚至算出 0。因為 $f(1+h)$ 和 $f(1)$ 太接近，電腦的小數位數不夠，相減時有效數字都消掉了。

所以實務上 h 取 $10^{-5}\sim10^{-6}$ 就夠了。

## 小結

- 極限問「靠近時往哪裡去」，左右都要看。
- 導數 = 差商在 h → 0 的極限，圖形上是切線斜率。
- 沒有公式時，可以用中央差分從數據估變化率。
- 導數要附單位、說出時刻。
