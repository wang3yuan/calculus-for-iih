## 1. 差商就是割線斜率

導數的定義是

$$f'(a)=\lim_{h\to 0}\frac{f(a+h)-f(a)}{h}$$

分子是函數值的**變化量**，分母是 x 的變化量，整個分數就是「曲線上兩點連線（割線）的斜率」。

我們用 $f(x)=x^2$，看 $a=3$ 時，h 越來越小，這個斜率變成多少。

```python
a = 3
for h in [1, 0.5, 0.1, 0.01, 0.001, 0.0001]:
    slope = ((a + h)**2 - a**2) / h        # 差商 = 割線斜率
    print(f"h = {h:<7}  割線斜率 = {slope:.4f}")
```

h 越小，斜率越靠近 **6**。這和課堂手算的結果一致：

$$\frac{(3+h)^2-3^2}{h}=\frac{6h+h^2}{h}=6+h\ \to\ 6$$

所以 $f'(3)=6$。

## 2. 看圖：割線慢慢變成切線

下圖畫出 $y=x^2$、點 $(3,9)$，以及 h 取不同值時的割線。h 越小，割線越貼近（綠色）切線。

```python
f = lambda x: x**2
a = 3
x = np.linspace(0.5, 5, 300)

plt.figure(figsize=(7, 5))
plt.plot(x, f(x), color="#0B2E33", lw=3, label="y = x²")
for h, c in [(2, "#E8A33D"), (1, "#D9534F"), (0.3, "#028090")]:
    m = (f(a + h) - f(a)) / h
    plt.plot(x, f(a) + m * (x - a), color=c, lw=1.8, label=f"secant, h = {h}, slope = {m:.2f}")
    plt.plot([a + h], [f(a + h)], "o", color=c)
plt.plot(x, f(a) + 6 * (x - a), color="green", lw=2.5, ls="--", label="tangent, slope = 6")
plt.plot([a], [f(a)], "ko")
plt.xlim(0.5, 5); plt.ylim(0, 20)
plt.xlabel("x"); plt.ylabel("y"); plt.grid(alpha=0.3); plt.legend()
plt.show()
```

## 5. 導數：把 h 變得非常小

有了 `slope`，我們可以直接用很小的 h 當作導數的近似。下面比較 $f(x)=x^3$ 在不同 x 的**電腦算的結果**和**公式** $3x^2$。

```python
def numeric_derivative(f, a, h=1e-6):
    """用很小的 h 近似導數（老師提供）。"""
    return (f(a + h) - f(a)) / h

g = lambda x: x**3
print("  x    電腦算的   公式 3x²")
for a in [-2, -1, 0, 1, 2, 3]:
    print(f"{a:>3}   {numeric_derivative(g, a):>8.3f}   {3 * a**2:>8}")
```

兩欄幾乎一模一樣，這就是「導數的公式是對的」的數值證據。

## 7. 情境：體溫的變化率

某位病人接受治療後，體溫 T（°C）隨時間 t（小時）變化，示意模型為

$$T(t)=36.5+1.2t-0.15t^2$$

**導數 $T'(t)$ 就是體溫的變化率**，單位是 °C／小時。先看圖。

```python
T = lambda t: 36.5 + 1.2 * t - 0.15 * t**2
t = np.linspace(0, 8, 200)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(t, T(t), color="#028090", lw=3)
ax[0].set_xlabel("t (hours)"); ax[0].set_ylabel("T (°C)"); ax[0].set_title("body temperature T(t)"); ax[0].grid(alpha=0.3)
ax[1].plot(t, [numeric_derivative(T, s) for s in t], color="#D9534F", lw=3)
ax[1].axhline(0, color="gray", lw=1)
ax[1].set_xlabel("t (hours)"); ax[1].set_ylabel("T' (°C/hour)"); ax[1].set_title("rate of change T'(t)"); ax[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 11. h 越小越好嗎？

數學上 h → 0 越小越準，但電腦存小數有**誤差**。看看 h 小到極端會發生什麼事：

```python
a = 3
print("   h        割線斜率   誤差（和 6 相差多少）")
for k in range(1, 17):
    h = 10.0**(-k)
    s = ((a + h)**2 - a**2) / h
    print(f"1e-{k:<2}  {s:>12.6f}   {abs(s - 6):.2e}")
```

前面 h 變小，誤差也變小；但到了 $10^{-8}$ 之後誤差反而變大，$10^{-16}$ 時甚至算出 0。原因是 $(3+h)^2$ 和 $3^2$ 太接近，電腦的小數位數不夠，相減時前面的位數都消掉了。

**所以實務上不是 h 越小越好**，通常取 $10^{-5}\sim10^{-6}$ 就夠用了。這也是為什麼我們前面把 `numeric_derivative` 的預設 h 設成 $10^{-6}$。
